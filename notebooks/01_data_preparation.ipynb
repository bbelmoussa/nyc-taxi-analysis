### **Categorical variables**

`date` contains **366 unique values**, ranging from **01-01-2024** to **31-12-2024**, covering the entire year 2024, including the leap day.

`hour` contains **24 unique values**, corresponding to the hours of the day.

`PU_Borough` and `DO_Borough` contain the same **7 geographical categories**: **Bronx, Brooklyn, Manhattan, Queens, Staten Island, EWR, and Unknown**.

`payment_type` contains payment categories defined as:

* 0 = Flex Fare trip
* 1 = Credit card
* 2 = Cash
* 3 = No charge
* 4 = Dispute
* 5 = Unknown
* 6 = Voided trip

However, the descriptive statistics show that the current dataset contains values only from **0 to 4**.

### **Discrete numerical variables**

`passenger_count` represents the number of passengers associated with each observation. The known values range from **1 to 5**, with an additional **Unknown** category.

`trip_count` represents the number of trips included in each observation, ranging from **1 to 7,242**, with a **median of 3 trips per observation**.

**The DataFrame does not present any missing or duplicated values, and no negative values were identified in the numerical variables.**

### **Aggregated-data consideration and grain definition**

The dataset documentation and the structure of the variables provide strong evidence that the data are **aggregated rather than individual trip-level records**. The numerical variables have different ranges, and most contain very large values compared with their mean and median. For example, the maximum `duration_sum` of 112,972.70 minutes would be unrealistic for a single trip, but it can be plausible as an aggregate value representing multiple trips. The large differences between the mean, median, and maximum also indicate that several numerical variables are **right-skewed**.

The dataset can be divided into two main parts. The first consists of **low-cardinality categorical, temporal, and discrete variables**, such as `date`, `hour`, `passenger_count`, the pickup and drop-off boroughs (`PU_Borough` and `DO_Borough`), and `payment_type`. These variables define the dimensions used to distinguish observations, and their combination uniquely identifies each row.

The second part consists of **numerical aggregate measures**, including `trip_count`, `trip_distance_sum`, `duration_sum`, `fare_amount_sum`, and `total_amount_sum`. These variables show strong positive linear correlations with `trip_count`, indicating that their values generally increase as the number of trips represented by an observation increases.

The grain is therefore **one row per unique combination of** `date`, `hour`, `passenger_count`, `PU_Borough`, `DO_Borough`, and `payment_type`, with the remaining numerical variables representing aggregated measures for that group.

```python
# temporal variables Conversion
df["date"] = pd.to_datetime(df["date"])
df["hour"] = pd.to_datetime(df["hour"], format="%H:%M").dt.time
```

```python
# categorical variables Conversion
categorical_cols = [
    "PU_Borough",
    "DO_Borough",
    "passenger_count"
]
df[categorical_cols] = df[categorical_cols].astype("category")
```

```python
# discrete numerical variable Conversion
df["trip_count"] = df["trip_count"].astype(int)
```

```python
# Data info Inspection
df.info()
```

```python
# Data grain Definition
grain_cols = [
    "date",
    "hour",
    "passenger_count",
    "PU_Borough",
    "DO_Borough",
    "payment_type"
]
```

```python
# grain Validation
df.duplicated(subset=grain_cols).sum()
```

```python
# grain Validation
df[grain_cols].nunique()
```

```python
# chronological sorting Inspection
df.groupby("date")["hour"].apply(lambda x: x.is_monotonic_increasing).value_counts()
```

```python
# aggregated Data Logical Consistency Inspection
calculated_total = (
    df["fare_amount_sum"]
    + df["extra_sum"]
    + df["mta_tax_sum"]
    + df["tip_amount_sum"]
    + df["tolls_amount_sum"]
    + df["improvement_surcharge_sum"]
    + df["congestion_surcharge_sum"]
    + df["airport_fee_sum"]
)

difference = calculated_total - df["total_amount_sum"]

print("Maximum absolute difference:", difference.abs().max())
print("Rows with difference > 0.01:", (difference.abs() > 0.01).sum())
```

## **Final Data Preparation Conclusion**

After converting the variables to their appropriate data types, the dataset was rechecked for **grain consistency, temporal consistency, and numerical consistency**.

The **six variables defining the observation grain** (`date`, `hour`, `passenger_count`, `PU_Borough`, `DO_Borough`, and `payment_type`) uniquely identify each row.

The **temporal checks** confirmed that the dates and hours are valid and that the hours are chronologically ordered within each day.

The **numerical consistency checks** also confirmed that the aggregate monetary components are consistent with `total_amount_sum`, with no meaningful discrepancies identified.

Together with the previous checks for **missing values, duplicates, invalid categories, and negative values**, these results indicate that the dataset is **internally consistent and ready for exploratory analysis**.

The cleaned dataset will now be exported and used as the input for the subsequent analysis notebooks.
